```python
def univariate_summary(series):
    return pd.Series({
        'count': series.count(),
        'mean': series.mean(),
        'median': series.median(),
        'mode': series.mode().iloc[0] if not series.mode().empty else np.nan,
        'std': series.std(),
        'variance': series.var(),
        'min': series.min(),
        'max': series.max(),
        'range': series.max() - series.min(),
        'Q1': series.quantile(0.25),
        'Q3': series.quantile(0.75),
        'IQR': series.quantile(0.75) - series.quantile(0.25),
        'skewness': series.skew(),
        'kurtosis': series.kurt()
    })

univariate_stats = pd.DataFrame({col: univariate_summary(df[col]) for col in numeric_cols_for_stats}).T
univariate_stats
```

### 9. K-Means Clustering

Standardizing numeric features, picking k via elbow + silhouette,
then fitting K-Means and projecting into 2D with PCA.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

cluster_features = ['AtBat', 'Hits', 'HmRun', 'Runs', 'RBI', 'Walks',
                     'Years', 'CAtBat', 'CHits', 'CHmRun', 'CRuns', 'CRBI',
                     'CWalks', 'PutOuts', 'Assists', 'Errors', 'Salary']

X = df[cluster_features].copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled.shape
```

```python
inertias = []
sil_scores = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(list(K_range), inertias, marker='o')
axes[0].set_xlabel('Number of clusters (k)')
axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('Elbow Method')

axes[1].plot(list(K_range), sil_scores, marker='o', color='darkorange')
axes[1].set_xlabel('Number of clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Analysis')
plt.tight_layout()
plt.show()

best_k = list(K_range)[int(np.argmax(sil_scores))]
print(f"Best k by silhouette score: {best_k}")
```

```python
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
cluster_profile = df.groupby('KMeans_Cluster')[cluster_features].mean()
cluster_profile
```

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
df['PCA1'], df['PCA2'] = X_pca[:, 0], X_pca[:, 1]

plt.figure(figsize=(9, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='KMeans_Cluster', data=df, palette='Set2', s=60)
plt.title(f"K-Means Clusters (k={best_k}) visualized via PCA")
plt.show()

print(f"Explained variance ratio (PC1, PC2): {pca.explained_variance_ratio_}")
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='KMeans_Cluster', y='Salary', data=df)
plt.title("Salary distribution by K-Means cluster")
plt.show()
```

### 10. Hierarchical Clustering

Building a Ward-linkage dendrogram on the same standardized data,
cutting at the same k, and comparing with K-Means results.

```python
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.cluster import AgglomerativeClustering

Z = linkage(X_scaled, method='ward')

plt.figure(figsize=(16, 7))
dendrogram(Z, truncate_mode='lastp', p=30, leaf_rotation=90., leaf_font_size=10.)
plt.title("Hierarchical Clustering Dendrogram (Ward linkage, truncated)")
plt.xlabel("Sample index (or cluster size)")
plt.ylabel("Distance")
plt.axhline(y=Z[-(best_k-1), 2], color='red', linestyle='--',
            label=f'Cut for {best_k} clusters')
plt.legend()
plt.show()
```

```python
agg = AgglomerativeClustering(n_clusters=best_k, linkage='ward')
df['Hierarchical_Cluster'] = agg.fit_predict(X_scaled)
df['Hierarchical_Cluster'].value_counts().sort_index()
```

```python
plt.figure(figsize=(9, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='Hierarchical_Cluster', data=df, palette='Set1', s=60)
plt.title(f"Hierarchical Clusters (k={best_k}) visualized via PCA")
plt.show()
```

```python
hier_profile = df.groupby('Hierarchical_Cluster')[cluster_features].mean()
hier_profile
```

```python
from sklearn.metrics import adjusted_rand_score, silhouette_score

ari = adjusted_rand_score(df['KMeans_Cluster'], df['Hierarchical_Cluster'])
sil_kmeans = silhouette_score(X_scaled, df['KMeans_Cluster'])
sil_hier = silhouette_score(X_scaled, df['Hierarchical_Cluster'])

print(f"Adjusted Rand Index (KMeans vs Hierarchical): {ari:.3f}")
print(f"Silhouette Score - K-Means:      {sil_kmeans:.3f}")
print(f"Silhouette Score - Hierarchical: {sil_hier:.3f}")

pd.crosstab(df['KMeans_Cluster'], df['Hierarchical_Cluster'],
            rownames=['KMeans'], colnames=['Hierarchical'])
```

### 12. Principal Component Analysis (PCA)

```python
features = [c for c in hit_raw.columns if c not in ['Salary','League','Division','NewLeague']]
X = StandardScaler().fit_transform(obs[features])
pca = PCA().fit(X)
scores = pca.transform(X)

evr = pca.explained_variance_ratio_
cum = np.cumsum(evr)
pd.DataFrame({'PC':[f'PC{i+1}' for i in range(len(evr))],
              'Eigenvalue':pca.explained_variance_.round(3),
              'Variance %':(evr*100).round(2),
              'Cumulative %':(cum*100).round(2)}).head(8)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14,5))
ax[0].bar(range(1,len(evr)+1), pca.explained_variance_, color='steelblue')
ax[0].axhline(1, color='red', ls='--', label='Kaiser rule (eigenvalue = 1)')
ax[0].set_title('Scree plot'); ax[0].set_xlabel('Component'); ax[0].set_ylabel('Eigenvalue'); ax[0].legend()
ax[1].plot(range(1,len(cum)+1), cum*100, 'o-', color='seagreen')
ax[1].axhline(90, color='red', ls='--', label='90% threshold')
ax[1].set_title('Cumulative explained variance'); ax[1].set_xlabel('Number of components'); ax[1].set_ylabel('%'); ax[1].legend()
plt.tight_layout(); plt.show()
print("Components needed for 80%:", np.argmax(cum>=0.80)+1, "| for 90%:", np.argmax(cum>=0.90)+1)
print("Components with eigenvalue > 1:", (pca.explained_variance_>1).sum())
```

```python
loadings = pd.DataFrame(pca.components_[:3].T, index=features, columns=['PC1','PC2','PC3']).round(2)
plt.figure(figsize=(6,7))
sns.heatmap(loadings, annot=True, cmap='coolwarm', center=0)
plt.title('Loadings of the first 3 components'); plt.show()
loadings
```

PC1 = career experience, PC2 = current-season hitting, PC3 = fielding.

### 13. Singular Value Decomposition (SVD)
X = U Σ Vᵀ, and the PCA eigenvalues equal σ² / (n-1).

```python
U, S, Vt = np.linalg.svd(X, full_matrices=False)
n_obs = X.shape[0]

check = pd.DataFrame({'Singular value': S[:5],
                      'sigma^2/(n-1)': S[:5]**2 / (n_obs - 1),
                      'PCA eigenvalue': pca.explained_variance_[:5]}, index=[f'PC{i+1}' for i in range(5)]).round(4)
same_scores = np.allclose(np.abs(U * S), np.abs(scores))
print("PCA scores == U*Sigma (up to sign):", same_scores)
check
```

```python
ks = np.arange(1, len(S) + 1)
rel_err = np.array([np.linalg.norm(X - (U[:, :k] * S[:k]) @ Vt[:k])**2 / np.linalg.norm(X)**2 for k in ks])

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].plot(ks, S, 'o-', color='steelblue'); ax[0].set_title('Singular values'); ax[0].set_xlabel('Index'); ax[0].set_ylabel('sigma')
ax[1].plot(ks, (1 - rel_err) * 100, 'o-', color='seagreen')
ax[1].set_title('Information retained by rank-k reconstruction'); ax[1].set_xlabel('Rank k'); ax[1].set_ylabel('% of squared norm retained')
plt.tight_layout(); plt.show()
pd.DataFrame({'k': [1, 2, 3, 5, 8], 'Relative reconstruction error': rel_err[[0, 1, 2, 4, 7]].round(3),
              'Retained %': ((1 - rel_err[[0, 1, 2, 4, 7]]) * 100).round(1)})
```

SVD matches PCA exactly; a rank-3 reconstruction keeps about 82% of the information.

### 14. Factor Analysis and Intrinsic Dimensionality

```python
p_vars = X.shape[1]
R = np.corrcoef(X, rowvar=False)
chi2_stat = -(n_obs - 1 - (2 * p_vars + 5) / 6) * np.log(np.linalg.det(R))
dof = p_vars * (p_vars - 1) / 2
p_bart = stats.chi2.sf(chi2_stat, dof)
print(f"Bartlett chi-square = {chi2_stat:.1f}, dof = {int(dof)}, p-value = {'< 1e-300' if p_bart == 0 else format(p_bart, '.3g')}")
```

```python
eig_real = np.sort(np.linalg.eigvalsh(R))[::-1]

kaiser = int((eig_real > 1).sum())
var80 = int(np.argmax(np.cumsum(eig_real) / eig_real.sum() >= 0.80) + 1)
rng = np.random.default_rng(0)
sim = np.array([np.sort(np.linalg.eigvalsh(np.corrcoef(rng.standard_normal((n_obs, p_vars)), rowvar=False)))[::-1]
                for _ in range(300)])
thr = np.percentile(sim, 95, axis=0)
parallel = int((eig_real > thr).sum())
fa_ll = [cross_val_score(FactorAnalysis(n_components=k, random_state=0), X, cv=5).mean() for k in range(1, 9)]

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].plot(range(1, p_vars + 1), eig_real, 'o-', label='Real data')
ax[0].plot(range(1, p_vars + 1), thr, 's--', color='red', label='Random data (95th pct)')
ax[0].axhline(1, color='grey', ls=':'); ax[0].set_title('Scree plot with parallel analysis')
ax[0].set_xlabel('Component'); ax[0].set_ylabel('Eigenvalue'); ax[0].legend()
ax[1].plot(range(1, 9), fa_ll, 'o-', color='darkorange'); ax[1].set_title('Factor Analysis: cross-validated log-likelihood')
ax[1].set_xlabel('Number of factors'); ax[1].set_ylabel('Mean CV log-likelihood')
plt.tight_layout(); plt.show()

pd.Series({'Kaiser (eigenvalue > 1)': kaiser, '80% variance': var80, 'Parallel analysis': parallel}, name='Suggested dimensions')
```

```python
fa3 = FactorAnalysis(n_components=3, rotation='varimax', random_state=0).fit(X)
fa_load = pd.DataFrame(fa3.components_.T, index=features, columns=['Factor 1', 'Factor 2', 'Factor 3'])
uniq = pd.Series(fa3.noise_variance_, index=features, name='Uniqueness')

plt.figure(figsize=(6, 7))
sns.heatmap(fa_load, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Varimax-rotated factor loadings'); plt.show()
pd.concat([fa_load.round(2), uniq.round(2)], axis=1)
```

Kaiser, 80% variance and parallel analysis all give 3 dimensions. Factors: career volume, current-season hitting, power hitting (signs are arbitrary).

### 15. Non-linear methods: MDS, Isomap, t-SNE and Self-Organizing Map

```python
embeddings = {
    'PCA (linear baseline)': PCA(n_components=2).fit_transform(X),
    'MDS': MDS(n_components=2, n_init=4, random_state=0).fit_transform(X),
    'Isomap (k=10)': Isomap(n_neighbors=10, n_components=2).fit_transform(X),
    't-SNE (perplexity=30)': TSNE(n_components=2, perplexity=30, init='pca', random_state=0).fit_transform(X),
}

fig, axes = plt.subplots(2, 2, figsize=(14, 11))
for ax, (name, Z) in zip(axes.ravel(), embeddings.items()):
    sc_ = ax.scatter(Z[:, 0], Z[:, 1], c=obs['Salary_log'], cmap='plasma', alpha=0.8, s=30)
    ax.set_title(name)
fig.colorbar(sc_, ax=axes, label='log(Salary)', shrink=0.8)
plt.show()

trust = pd.Series({name: trustworthiness(X, Z, n_neighbors=10) for name, Z in embeddings.items()}, name='Trustworthiness').round(3)
trust
```

Trustworthiness is about 0.90 or higher for all methods (t-SNE best at 0.97); PCA is nearly as good, so there is little non-linear structure.

```python
def train_som(data, grid_x=8, grid_y=8, iters=8000, lr0=0.5, seed=0):
    rng = np.random.default_rng(seed)
    W = rng.standard_normal((grid_x, grid_y, data.shape[1])) * 0.1
    gx, gy = np.meshgrid(np.arange(grid_x), np.arange(grid_y), indexing='ij')
    sigma0 = max(grid_x, grid_y) / 2
    for t in range(iters):
        x = data[rng.integers(len(data))]
        d = ((W - x) ** 2).sum(axis=2)
        bi, bj = np.unravel_index(d.argmin(), d.shape)
        lr = lr0 * np.exp(-3 * t / iters)
        sigma = sigma0 * np.exp(-3 * t / iters) + 0.5
        h = np.exp(-((gx - bi) ** 2 + (gy - bj) ** 2) / (2 * sigma ** 2))
        W += lr * h[..., None] * (x - W)
    return W

som = train_som(X)
gx_, gy_ = som.shape[:2]

bmus = np.array([np.unravel_index(((som - x) ** 2).sum(axis=2).argmin(), (gx_, gy_)) for x in X])
q_err = np.mean([np.sqrt(((som - x) ** 2).sum(axis=2).min()) for x in X])
node_sal = np.full((gx_, gy_), np.nan); node_cnt = np.zeros((gx_, gy_), dtype=int)
for i in range(gx_):
    for j in range(gy_):
        m = (bmus[:, 0] == i) & (bmus[:, 1] == j); node_cnt[i, j] = m.sum()
        if m.any(): node_sal[i, j] = obs['Salary_log'].values[m].mean()

umat = np.zeros((gx_, gy_))
for i in range(gx_):
    for j in range(gy_):
        nb_d = [np.linalg.norm(som[i, j] - som[a, b]) for a, b in [(i-1, j), (i+1, j), (i, j-1), (i, j+1)]
                if 0 <= a < gx_ and 0 <= b < gy_]
        umat[i, j] = np.mean(nb_d)

fig, ax = plt.subplots(1, 3, figsize=(18, 5))
im0 = ax[0].imshow(umat, cmap='bone_r'); ax[0].set_title('U-matrix (distance between neighbouring nodes)'); plt.colorbar(im0, ax=ax[0])
im1 = ax[1].imshow(node_cnt, cmap='Blues'); ax[1].set_title('Hit map (players per node)'); plt.colorbar(im1, ax=ax[1])
im2 = ax[2].imshow(node_sal, cmap='plasma'); ax[2].set_title('Mean log(Salary) per node'); plt.colorbar(im2, ax=ax[2])
plt.tight_layout(); plt.show()
print(f"Quantization error: {q_err:.3f} | occupied nodes: {(node_cnt > 0).sum()} of {gx_ * gy_}")
```

The SOM salary map shows a low-to-high gradient although salary was not used in training.

#### 16.1 Evaluation and accuracy

```python
Xtr_s, Xte_s, ytr_s, yte_s = train_test_split(obs[[predictor]], obs['Salary_log'], test_size=0.25, random_state=42)
from sklearn.linear_model import LinearRegression
lin = LinearRegression().fit(Xtr_s, ytr_s)
baseline = np.full(len(yte_s), ytr_s.mean())

def report(y_true, y_pred, n_pred=1):
    n_ = len(y_true); r2_ = r2_score(y_true, y_pred)
    sal_true, sal_pred = np.exp(y_true), np.exp(y_pred)
    return {'R2': r2_, 'Adj R2': 1 - (1 - r2_) * (n_ - 1) / (n_ - n_pred - 1),
            'RMSE (log scale)': np.sqrt(mean_squared_error(y_true, y_pred)),
            'MAE (log scale)': mean_absolute_error(y_true, y_pred),
            'MAE ($K)': mean_absolute_error(sal_true, sal_pred),
            'Within 50% of actual salary': np.mean(np.abs(sal_pred - sal_true) / sal_true <= 0.5)}

metrics = pd.DataFrame({'Train': report(ytr_s, lin.predict(Xtr_s)),
                        'Test': report(yte_s, lin.predict(Xte_s)),
                        'Test: baseline (mean)': report(yte_s, baseline, n_pred=0)}).round(3)

from sklearn.model_selection import KFold
cv_r2 = cross_val_score(LinearRegression(), obs[[predictor]], obs['Salary_log'],
                        cv=KFold(5, shuffle=True, random_state=42), scoring='r2')
print("5-fold CV R2:", cv_r2.round(3), "| mean = %.3f, std = %.3f" % (cv_r2.mean(), cv_r2.std()))
metrics
```

```python
pred_te = np.exp(lin.predict(Xte_s)); act_te = np.exp(yte_s)
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].scatter(act_te, pred_te, alpha=0.7); lim = [0, max(act_te.max(), pred_te.max())]
ax[0].plot(lim, lim, 'r--', label='Perfect prediction'); ax[0].set_xlabel('Actual salary ($K)'); ax[0].set_ylabel('Predicted salary ($K)')
ax[0].set_title('Predicted vs actual (test set)'); ax[0].legend()
ax[1].hist(yte_s - lin.predict(Xte_s), bins=15, color='steelblue', edgecolor='white')
ax[1].set_title('Test residuals (log scale)'); ax[1].set_xlabel('Actual - predicted')
plt.tight_layout(); plt.show()
```

- Train (0.39), test (0.38) and CV (about 0.35) R-squared are close, so no over-fitting; the mean baseline has R-squared of about 0.
- Typical error is a factor of about 1.7 to 1.9 (test RMSE 0.66 vs 0.84 for the baseline).
- The model explains the trend but not individual salaries.

### 17. Classification: Logistic Regression
Target: high vs low salary (above the median).

```python
med = obs['Salary'].median()
obs['HighSalary'] = (obs['Salary'] > med).astype(int)
print("Median salary (K$):", med)
print(obs['HighSalary'].value_counts())

cls_feats = features + ['League','Division']
Xc = obs[cls_feats].copy()
Xc['League'] = (Xc['League']=='N').astype(int)
Xc['Division'] = (Xc['Division']=='W').astype(int)
yc = obs['HighSalary']

Xtr, Xte, ytr, yte = train_test_split(Xc, yc, test_size=0.25, stratify=yc, random_state=42)
sc = StandardScaler().fit(Xtr)
clf = LogisticRegression(max_iter=1000)
clf.fit(sc.transform(Xtr), ytr)
yp = clf.predict(sc.transform(Xte)); yprob = clf.predict_proba(sc.transform(Xte))[:,1]
print(f"Test accuracy: {accuracy_score(yte,yp):.3f} | ROC-AUC: {roc_auc_score(yte,yprob):.3f}")
print(classification_report(yte, yp, target_names=['Low','High']))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13,5))
ConfusionMatrixDisplay(confusion_matrix(yte,yp), display_labels=['Low','High']).plot(ax=ax[0], cmap='Blues', colorbar=False)
ax[0].set_title('Confusion matrix (test set)')
fpr, tpr, _ = roc_curve(yte, yprob)
ax[1].plot(fpr, tpr, label=f'AUC = {roc_auc_score(yte,yprob):.2f}'); ax[1].plot([0,1],[0,1],'k--')
ax[1].set_xlabel('False positive rate'); ax[1].set_ylabel('True positive rate'); ax[1].set_title('ROC curve'); ax[1].legend()
plt.tight_layout(); plt.show()

cv = cross_val_score(LogisticRegression(max_iter=1000),
                     StandardScaler().fit_transform(Xc), yc,
                     cv=StratifiedKFold(5, shuffle=True, random_state=42), scoring='accuracy')
print("5-fold CV accuracy:", cv.round(3), "| mean =", cv.mean().round(3))
```

```python
coefs = pd.Series(clf.coef_[0], index=cls_feats).sort_values()
plt.figure(figsize=(8,6))
coefs.plot(kind='barh', color=np.where(coefs>0,'seagreen','crimson'))
plt.title('Logistic regression coefficients (standardized features)'); plt.axvline(0,color='k'); plt.show()
coefs.round(2).sort_values(key=abs, ascending=False).head(6)
```

Test accuracy about 80%, AUC about 0.88, CV about 0.82. Five PCs give nearly the same accuracy.
