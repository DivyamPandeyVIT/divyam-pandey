#### 6.1 Detailed numeric summaries

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_like = ['League', 'Division', 'NewLeague']
numeric_cols_for_stats = [c for c in numeric_cols if c not in categorical_like]
numeric_cols_for_stats
```

```python
def univariate_summary(series):
    return pd.Series({
        'count': series.count(),
        'mean': series.mean(),
        'median': series.median(),
        'mode': series.mode().iloc[0] if not series.mode().empty else np.nan,
        'std': series.std(),
        'variance': series.var(),
        'min': series.min(),
        'max': series.max(),
        'range': series.max() - series.min(),
        'Q1': series.quantile(0.25),
        'Q3': series.quantile(0.75),
        'IQR': series.quantile(0.75) - series.quantile(0.25),
        'skewness': series.skew(),
        'kurtosis': series.kurt()
    })

univariate_stats = pd.DataFrame({col: univariate_summary(df[col]) for col in numeric_cols_for_stats}).T
univariate_stats
```

```python
for col in numeric_cols_for_stats:
    sk = df[col].skew()
    if sk > 1:
        tag = 'highly right-skewed'
    elif sk > 0.5:
        tag = 'moderately right-skewed'
    elif sk < -1:
        tag = 'highly left-skewed'
    elif sk < -0.5:
        tag = 'moderately left-skewed'
    else:
        tag = 'approximately symmetric'
    print(f"{col:10s} skew={sk:6.2f}  -> {tag}")
```

### 9. K-Means Clustering

Standardizing numeric features, picking k via elbow + silhouette,
then fitting K-Means and projecting into 2D with PCA.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

cluster_features = ['AtBat', 'Hits', 'HmRun', 'Runs', 'RBI', 'Walks',
                     'Years', 'CAtBat', 'CHits', 'CHmRun', 'CRuns', 'CRBI',
                     'CWalks', 'PutOuts', 'Assists', 'Errors', 'Salary']

X = df[cluster_features].copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled.shape
```

```python
inertias = []
sil_scores = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(list(K_range), inertias, marker='o')
axes[0].set_xlabel('Number of clusters (k)')
axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('Elbow Method')

axes[1].plot(list(K_range), sil_scores, marker='o', color='darkorange')
axes[1].set_xlabel('Number of clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Analysis')
plt.tight_layout()
plt.show()

best_k = list(K_range)[int(np.argmax(sil_scores))]
print(f"Best k by silhouette score: {best_k}")
```

```python
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
cluster_profile = df.groupby('KMeans_Cluster')[cluster_features].mean()
cluster_profile
```

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
df['PCA1'], df['PCA2'] = X_pca[:, 0], X_pca[:, 1]

plt.figure(figsize=(9, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='KMeans_Cluster', data=df, palette='Set2', s=60)
plt.title(f"K-Means Clusters (k={best_k}) visualized via PCA")
plt.show()

print(f"Explained variance ratio (PC1, PC2): {pca.explained_variance_ratio_}")
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='KMeans_Cluster', y='Salary', data=df)
plt.title("Salary distribution by K-Means cluster")
plt.show()
```

### 10. Hierarchical Clustering

Building a Ward-linkage dendrogram on the same standardized data,
cutting at the same k, and comparing with K-Means results.

```python
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.cluster import AgglomerativeClustering

Z = linkage(X_scaled, method='ward')

plt.figure(figsize=(16, 7))
dendrogram(Z, truncate_mode='lastp', p=30, leaf_rotation=90., leaf_font_size=10.)
plt.title("Hierarchical Clustering Dendrogram (Ward linkage, truncated)")
plt.xlabel("Sample index (or cluster size)")
plt.ylabel("Distance")
plt.axhline(y=Z[-(best_k-1), 2], color='red', linestyle='--',
            label=f'Cut for {best_k} clusters')
plt.legend()
plt.show()
```

```python
agg = AgglomerativeClustering(n_clusters=best_k, linkage='ward')
df['Hierarchical_Cluster'] = agg.fit_predict(X_scaled)
df['Hierarchical_Cluster'].value_counts().sort_index()
```

```python
plt.figure(figsize=(9, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='Hierarchical_Cluster', data=df, palette='Set1', s=60)
plt.title(f"Hierarchical Clusters (k={best_k}) visualized via PCA")
plt.show()
```

```python
hier_profile = df.groupby('Hierarchical_Cluster')[cluster_features].mean()
hier_profile
```

```python
from sklearn.metrics import adjusted_rand_score, silhouette_score

ari = adjusted_rand_score(df['KMeans_Cluster'], df['Hierarchical_Cluster'])
sil_kmeans = silhouette_score(X_scaled, df['KMeans_Cluster'])
sil_hier = silhouette_score(X_scaled, df['Hierarchical_Cluster'])

print(f"Adjusted Rand Index (KMeans vs Hierarchical): {ari:.3f}")
print(f"Silhouette Score - K-Means:      {sil_kmeans:.3f}")
print(f"Silhouette Score - Hierarchical: {sil_hier:.3f}")

pd.crosstab(df['KMeans_Cluster'], df['Hierarchical_Cluster'],
            rownames=['KMeans'], colnames=['Hierarchical'])
```
