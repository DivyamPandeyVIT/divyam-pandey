# EDA Project - Phase 1 (Hitters Dataset)

### 1. Load dataset

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Hitters.csv"
df = pd.read_csv(url)
df.head()
```

```
df.shape
```

```
df.info()
```

### 2. Basic statistical analysis

```
df.describe()
```

```
df.describe(include='all')
```

### 3. Handling missing data

```
df.isnull().sum()
```

```
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df.isnull().sum()
```

### 4. Data cleaning

```
df.duplicated().sum()
```

```
df = df.drop_duplicates()
df.columns = df.columns.str.strip()
```

```
df['League'].unique(), df['Division'].unique(), df['NewLeague'].unique()
```

### 5. Data transformation

```
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['League'] = le.fit_transform(df['League'])
df['Division'] = le.fit_transform(df['Division'])
df['NewLeague'] = le.fit_transform(df['NewLeague'])

df['Salary_log'] = np.log(df['Salary'] + 1)
df.head()
```

### 6. Univariate analysis

```
plt.hist(df['Salary'], bins=20)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.show()
```

```
sns.boxplot(x=df['Hits'])
plt.title("Hits Boxplot")
plt.show()
```

```
df['League'].value_counts().plot(kind='bar')
plt.title("League Count")
plt.show()
```

### 7. Bivariate analysis

```
plt.scatter(df['CAtBat'], df['Salary'])
plt.xlabel("CAtBat")
plt.ylabel("Salary")
plt.title("CAtBat vs Salary")
plt.show()
```

```
sns.boxplot(x='Division', y='Salary', data=df)
plt.title("Salary by Division")
plt.show()
```

```
plt.scatter(df['Hits'], df['Runs'])
plt.xlabel("Hits")
plt.ylabel("Runs")
plt.title("Hits vs Runs")
plt.show()
```

### 8. Multivariate analysis

```
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```
sns.pairplot(df[['Salary', 'Hits', 'HmRun', 'Runs']])
plt.show()
```

```
sns.scatterplot(x='CAtBat', y='Salary', hue='Division', size='Years', data=df)
plt.title("CAtBat vs Salary by Division and Years")
plt.show()
```
